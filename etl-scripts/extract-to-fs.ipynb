```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
baseurl='https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_' 
#https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2025-01.parquet
urls = []
for i in range(2015,2025):
    for j in range(1,13):
        url = baseurl+str(i)+'-'+str(j).zfill(2)+'.parquet'
        urls.append(url)        
```

```python
import boto3
import urllib
from urllib import request     # For opening and reading URLs
from urllib import parse       # For parsing and encoding URLs and query strings
from urllib import error       # For handling exceptions like HTTPError, URLError
from urllib import robotparser # For parsing robots.txt files


s3 = boto3.client('s3')
bucket_name = 'nyc-yellow-taxi-parquetdata'

for url in urls:
    file_name = url.split('/')[-1]
    try:
        print(f"Downloading {url}...")
        response = request.urlopen(url)
        data = response.read()
        
        s3.put_object(Bucket=bucket_name, Key=file_name, Body=data)
        print(f"Uploaded {file_name} to S3")
    except error.URLError as e:
        print(f"URL Error: {e.reason}")
    except Exception as e:
        print(f"Unexpected error: {e}")
```
