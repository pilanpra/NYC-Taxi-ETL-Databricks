```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
baseurl='https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_' 
#https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2025-01.parquet
urls = []
for i in range(2022,2025):
    for j in range(1,13):
        url = baseurl+str(i)+'-'+str(j).zfill(2)+'.parquet'
        urls.append(url)        
```

```python
# Importing necessary libraries
import urllib
from urllib import request     # For opening and reading URLs
from urllib import parse       # For parsing and encoding URLs and query strings
from urllib import error       # For handling exceptions like HTTPError, URLError
from urllib import robotparser # For parsing robots.txt files
```

```python
import os

# Create the 'raw' folder if it doesn't exist
os.makedirs('raw', exist_ok=True)

# Download each file in the urls array
for url in urls:
    filename = os.path.join('raw', os.path.basename(url))
    try:
        print(f"Downloading {url}...")
        urllib.request.urlretrieve(url, filename)
        print(f"Saved to {filename}")
    except urllib.error.URLError as e:
        print(f"Failed to download {url}: {e}")
```
